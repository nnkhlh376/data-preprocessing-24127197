## 1. BoW và TF-IDF n-gram

**Lý thuyết — cần trình bày:**
- $\text{tfidf}(t,d) = tf(t,d) \cdot \log\frac{1+N}{1+df(t)} + 1$ (công thức mặc định của scikit-learn).
- Nêu rõ n ∈ {1,2,3} nghĩa là `ngram_range=(1,n)` hay `(n,n)`.
- Sparsity $= 1 - \frac{\text{nnz}}{N \times D}$.

> **[Code]** `CountVectorizer`, `TfidfVectorizer` với các n-gram; số chiều, sparsity; điểm Logistic Regression 5-fold CV.

## 3. Cosine similarity cùng lớp và khác lớp

**Lý thuyết — cần trình bày:**
- $\cos(u,v) = \frac{u \cdot v}{\|u\| \|v\|}$; cosine cao không tự chứng minh cùng cảm xúc (cùng chủ đề phim, nhiều từ chung).

## 4. t-SNE 2D và silhouette

**Lý thuyết — cần trình bày:**
- Silhouette $s = \frac{b - a}{\max(a,b)}$; **nêu rõ dùng nhãn nào**: nhãn thật (đo mức tách lớp) hay nhãn cụm.
- t-SNE chỉ để quan sát, không thay thế đánh giá phân loại.
