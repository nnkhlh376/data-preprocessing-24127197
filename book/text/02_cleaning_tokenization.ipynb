## 1. Pipeline chuẩn hóa

**Lý thuyết — cần trình bày:**
- Mục đích từng bước; IMDb có nhiều `<br />`, ít mention/hashtag (vẫn cài đặt và báo cáo tần suất).
- Quy tắc cho số: điểm như "7/10" mang thông tin cảm xúc → "theo điều kiện" nghĩa là gì trong bài này.
- Không làm mất phủ định: `not good`, `don't` không được biến thành `good`.

> **[Code]** Viết từng bước thành hàm riêng; ví dụ trước/sau cho vài review; đếm số lần mỗi bước tác động.

## 2. Tác động lên từ vựng và độ dài

**Lý thuyết — cần trình bày:**
- Tỷ lệ thay đổi từ vựng $= 1 - |V_{after}| / |V_{before}|$ (từ vựng sau / trước khi làm sạch) (gọi đúng tên: tỷ lệ giảm số dạng từ).

> **[Code]** Bảng $|V|$ và độ dài trung bình trước/sau từng bước; gán vào biến `cleaning_effect`.

> **[Nhận xét]** Làm sạch mạnh hơn có chắc tốt hơn không?

## 3. Bốn chiến lược tokenization

**Lý thuyết — cần trình bày:**
- Word-level (NLTK/spaCy), sentence-level (chia câu — đơn vị rất thưa), character-level, BPE (học các cặp ký tự hay gặp, ghép dần).
- **Định nghĩa OOV:** token trong test không có trong từ vựng học từ **train**; báo cáo theo loại (type) và theo lượt (token).
- BPE và character-level gần như không có OOV theo thiết kế; sentence-level gần như toàn OOV — các con số không tương đương trực tiếp.

> **[Code]** Cài đặt 4 tokenizer; BPE huấn luyện chỉ trên train; tính kích thước từ vựng, OOV trên test, độ dài token trung bình; gom vào biến `tokenization_results`.
