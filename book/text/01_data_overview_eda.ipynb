## 2. Độ dài văn bản và Mann–Whitney U

**Lý thuyết — cần trình bày:**
- Mann–Whitney U dựa trên thứ hạng: $U = n_1 n_2 + \frac{n_1(n_1+1)}{2} - R_1$; $H_0$: $P(X > Y) = 0.5$ — không mặc định là kiểm định trung bình hay trung vị.
- Kích thước hiệu ứng (rank-biserial) vì n lớn dễ cho p-value nhỏ.

> **[Code]** Tính số từ, số ký tự mỗi review; histogram/boxplot theo nhãn; Mann–Whitney U, gán vào biến `mwu_words`, `mwu_chars`.

## 4. Type-token ratio (TTR)

**Lý thuyết — cần trình bày:**
- $TTR = \frac{\#\text{types}}{\#\text{tokens}}$ (số loại từ / số token) — phụ thuộc độ dài văn bản, nên so sánh trên đoạn cùng độ dài hoặc trung bình theo review.

> **[Code]** Tính TTR cho toàn lớp và trung bình theo review; gán vào biến `ttr_results`.

> **[Nhận xét]** Lớp nào dùng từ đa dạng hơn? Có do độ dài khác nhau không?

## 5. Định luật Zipf

**Lý thuyết — cần trình bày:**
- $f(r) \propto r^{-s}$, tức $\log f = c - s \log r$; Zipf lý tưởng có $s \approx 1$.

> **[Code]** Vẽ log-log tần suất theo thứ hạng; hồi quy tuyến tính trên log-log để ước lượng $s$ và $R^2$ (gán vào biến `zipf_slope`, `zipf_r2`).
